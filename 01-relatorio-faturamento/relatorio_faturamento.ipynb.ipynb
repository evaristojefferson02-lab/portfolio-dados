```python
import pandas as pd

# Carregar os dados
dados = pd.read_csv("dados_vendas.csv")
print("✅ Dados carregados!")
dados
```

```python
# Calcular total por produto
dados["Faturamento"] = dados["Quantidade"] * dados["Valor_Unitario"]
dados
```

```python
# Resumo
total_geral = dados["Faturamento"].sum()
media = dados["Faturamento"].mean()

print("=" * 40)
print("RELATÓRIO DE FATURAMENTO")
print("=" * 40)
print(f"Total geral: R$ {total_geral:,.2f}")
print(f"Média por produto: R$ {media:,.2f}")
print("=" * 40)

# Mostrar produto mais vendido
mais_vendido = dados.loc[dados["Quantidade"].idxmax()]
print(f"Produto mais vendido: {mais_vendido['Produto']} — {mais_vendido['Quantidade']} unidades")
```

```python
# Salvar relatório pronto
dados.to_csv("resultado_final.csv", index=False)
print("✅ Arquivo 'resultado_final.csv' gerado com sucesso!")
```
